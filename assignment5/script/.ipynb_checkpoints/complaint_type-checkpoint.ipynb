```python
import pandas as pd 
import os
import matplotlib.pyplot as plt 

dt = pd.read_csv(os.path.join("..", "data", "cleaned_data.csv"), parse_dates=["Created Date", "Closed Date"])

dt_early_2024 = dt[(dt['Created Date'].dt.year == 2024 ) & (dt['Created Date'].dt.month.isin([1,2]))]

dt_june_july_2024 = dt[(dt['Created Date'].dt.year == 2024 ) & (dt['Created Date'].dt.month.isin([6,7]))]

most_complaint_type = dt_early_2024["Complaint Type"].value_counts().idxmax()

print(dt_early_2024.shape[0], "rows found")
print(most_complaint_type)

df_most_common_complaints = dt_early_2024[dt_early_2024["Complaint Type"] == most_complaint_type]

monthly_counts = df_most_common_complaints.groupby(df_most_common_complaints['Created Date'].dt.month).size()

print(monthly_counts)

plt.figure(figsize=(8,5))
monthly_counts.plot(kind='bar')
plt.xticks([0,1], ['January', 'February'], rotation = 0)
plt.show()
```
