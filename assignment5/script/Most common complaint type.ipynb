```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("cleaned_data.csv", parse_dates=["Created Date"])

df_early_2024 = df[(df["Created Date"].dt.year == 2024) & 
                   (df["Created Date"].dt.month.isin([1, 2]))]

most_common_type = df_early_2024['Complaint Type'].value_counts().idxmax()

df_most_common = df_early_2024[df_early_2024['Complaint Type'] == most_common_type]

monthly_counts = df_most_common.groupby(df_most_common['Created Date'].dt.month).size()

plt.figure(figsize=(8,5))
monthly_counts.plot(kind='bar', color='skyblue')
plt.xticks([0,1], ['January', 'February'], rotation=0)  # Label x-axis with month names
plt.title(f"Occurrences of '{most_common_type}' in Jan-Feb 2024")
plt.ylabel("Number of Complaints")
plt.xlabel("Month")
plt.show()
```
